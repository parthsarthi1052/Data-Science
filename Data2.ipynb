List,Tupple, Dictionary

```python
lst=[1,2,3,'hello','c',[2,3]]

print(lst)
# append()
lst.append([6,7,8])
print(lst)
# gets in as a list
```

```python
lst=[1,2,3,'hello','c',[2,3]]

# extend()
lst.extend([6,7,8])
print(lst)
# they all get in as separate elements in lst.
```

```python
lst=[1,2,3,'hello','c',[2,3]]

# insert
# lst.insert(index,value)
lst.insert(2,"value")
print(lst)
```

```python
# append()
# adds value to the last of the list.
lst=["Aryan","lost"]
lst.append("phone")
print(lst)
```

```python
# pop()
lst=["Aryan","Has","phone"]
print(lst)
lst.pop()
print(lst)
```

```python
# remove()
# removes the first occurence of that element
lst=[1,2,3,3,4,4,5]
lst.remove(3)
print(lst)
```

```python
# index
# returns the index of a particular value
lst=[1,2,3,4]
print(lst.index(3))
```

```python
# count
# gives the no of times that thing is coming
lst=["hi","hi","hi"]
print(lst.count("hi"))
```

```python
# sort()
lst=[6,5,3,7,1]
lst.sort()
print(lst)
```

```python
# sum,min,max,sorted
lst=[7,5,9,0,8,3,11]
print(min(lst))
print(max(lst))
print(sum(lst))
print(sorted(lst))
print(lst)
```
